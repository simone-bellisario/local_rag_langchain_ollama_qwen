# 2. Splitting the document into chunks

```python
#Using a semantic splitter for greater context awareness
```

```python
semantic_splitter = SemanticChunker(
    embeddings = embedding_model,
    breakpoint_threshold_type = 'gradient',
    breakpoint_threshold_amount = 80

)
```

```python
docs = semantic_splitter.split_documents(ai_act)
```

# 3. Creating a vector store

```python
persist_directory = os.path.join(os.getcwd(), "chroma_db")
```

```python
vectorstore = Chroma.from_documents(
    docs,
    embedding=embedding_model,
    persist_directory=persist_directory
)
```

```python
base_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 15}
)

cross_encoder = HuggingFaceCrossEncoder(model_name="BAAI/bge-reranker-base")
reranker = CrossEncoderReranker(model=cross_encoder, top_n=3)

retriever = ContextualCompressionRetriever(
    base_compressor=reranker,
    base_retriever=base_retriever
)
```

```python
system_message = """You are an assistant that answers questions about the EU Artificial Intelligence Act (Regulation (EU) 2024/1689).

Rules:
1. Use ONLY the context provided. Do not use outside knowledge.
2. If the context does not contain the answer, reply exactly: "I couldn't find this in the provided AI Act text."
3. Always cite the source of each claim in brackets, e.g. [Article 6], [Recital 47], [Annex III], using the labels shown in the context.
4. Use the Act's exact terminology (e.g. "provider", "deployer", "high-risk AI system", "general-purpose AI model") and do not substitute synonyms.
5. Be concise and precise. This is general information, not legal advice.
6. Answer in the same language as the question."""

human_message = """Context:
{context}

Question:
{question}

Answer:"""

prompt_template = ChatPromptTemplate.from_messages([
    ("system", system_message),
    ("human", human_message),
])
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt_template
    | llm
    | StrOutputParser()
)
```

```python
answer = rag_chain.invoke("What activities are considered an unacceptable risk? Make concrete examples.")
print(answer)
```

Context Precision

```python
import sys, types, importlib.util

try:
    import langchain_community.chat_models.vertexai
except ImportError:
    shim = types.ModuleType("langchain_community.chat_models.vertexai")
    shim.ChatVertexAI = type("ChatVertexAI", (), {})
    sys.modules[shim.__name__] = shim

from langchain_ollama import ChatOllama
from ragas import EvaluationDataset, evaluate
from ragas.llms import LangchainLLMWrapper
from ragas.metrics import LLMContextPrecisionWithReference, LLMContextRecall, Faithfulness
from ragas.run_config import RunConfig


def evaluate_with_ragas(test_set, retriever, rag_chain, judge_model="qwen2.5:1.5b"):
    judge = LangchainLLMWrapper(
        ChatOllama(model=judge_model, temperature=0, num_ctx=8192, reasoning=False)
    )

    rows = []
    for item in test_set:
        docs = retriever.invoke(item["question"])
        rows.append({
            "user_input": item["question"],
            "retrieved_contexts": [d.page_content for d in docs],
            "response": rag_chain.invoke(item["question"]),
            "reference": item["reference"],
        })

    result = evaluate(
        dataset=EvaluationDataset.from_list(rows),
        metrics=[LLMContextPrecisionWithReference(), LLMContextRecall(), Faithfulness()],
        llm=judge,
        run_config=RunConfig(timeout=600, max_workers=1),
    )
    return result.to_pandas()
```

```python
test_set = [
    {
        "question": "What is the maximum fine for violating the prohibited AI practices?",
        "reference": "Up to EUR 35,000,000 or 7% of total worldwide annual turnover, whichever is higher.",
    },
    # ... add more questions, with reference answers you checked against the official text
]

df = evaluate_with_ragas(test_set, retriever, rag_chain)

print(df.drop(columns=["retrieved_contexts"]))
print(df.mean(numeric_only=True))
```

Faithfulness

```python
from langchain_ollama import ChatOllama
from ragas import SingleTurnSample
from ragas.llms import LangchainLLMWrapper
from ragas.metrics import Faithfulness

# Judge model, created once and reusable
judge = LangchainLLMWrapper(
    ChatOllama(model="qwen2.5:1.5b", temperature=0, num_ctx=8192, reasoning=False)
)
faithfulness = Faithfulness(llm=judge)

query = "What are the obligations of providers of high-risk AI systems?"

# Run your real pipeline
retrieved_docs = [doc.page_content for doc in retriever.invoke(query)]
answer = rag_chain.invoke(query)

# Score the answer against the chunks it was based on
sample = SingleTurnSample(
    user_input=query,
    response=answer,
    retrieved_contexts=retrieved_docs,
)
score = await faithfulness.single_turn_ascore(sample)

print(f"Answer: {answer}\n")
print(f"Faithfulness: {score:.2f}")
```
